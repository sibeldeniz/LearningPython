## Part 3 · Variables and data types

📘 A **variable** is a name that holds a value. `=` means *"store the value on the right under the name on the left"* (it is not a maths equation). Python has four basic kinds of value:

| Type | Python name | Holds | Example |
|---|---|---|---|
| Text | `str` | letters in quotes | `"P-glycoprotein"` |
| Whole number | `int` | counts | `100` |
| Decimal number | `float` | measurements | `0.24` |
| True / False | `bool` | yes-or-no facts | `True` |

`type(x)` tells you which kind a value is. **Naming rules:** lower-case words joined by `_` (`rmsd_nm`), no spaces, no `-`, never start with a digit. Put the **unit in the name** (`simulation_time_ns`): it prevents costly mix-ups.

▶️ **Worked example 3** (the values are illustrative, not real results):

````python
protein_name = "P-glycoprotein"
simulation_time_ns = 100
rmsd_nm = 0.24
simulation_completed = True

print(protein_name, simulation_time_ns, rmsd_nm, simulation_completed)
print(type(protein_name))
print(type(simulation_time_ns))
print(type(rmsd_nm))
print(type(simulation_completed))
````

Expected output: the four values on one line, then `<class 'str'>`, `<class 'int'>`, `<class 'float'>`, `<class 'bool'>`.

**Updating a variable:** the right side is calculated first, then stored again under the same name.

````python
simulation_time_ns = simulation_time_ns + 50    # the old value (100) plus 50
print(simulation_time_ns)
````

## Part 4 · Arithmetic and scientific calculations

📘 Python is a calculator that remembers.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` `-` `*` | add, subtract, multiply | `6 * 7` | `42` |
| `/` | divide (always gives a decimal) | `7 / 2` | `3.5` |
| `//` | floor division (whole part) | `7 // 2` | `3` |
| `%` | remainder | `7 % 2` | `1` |
| `**` | power | `10 ** 3` | `1000` |

**Order of operations** is as in maths: brackets, then powers, then `*` `/`, then `+` `-`. So `2 + 3 * 4` is `14` but `(2 + 3) * 4` is `20`.
**Comparisons** give a bool: `==` equal, `!=` not equal, `<` `<=` `>` `>=`. (A single `=` stores a value; a double `==` asks a question.)
**Rounding:** `round(x, 2)` keeps 2 decimals. Computers store decimals approximately, so `0.24 * 10` can print as `2.4000000000000004`. Rounding for display is normal.

**Units you will need:** 1 ns = 1000 ps, and **1 nm = 10 Å**. So nm → Å means *multiply by 10*, and Å → nm means *divide by 10*.

▶️ **Worked example 4**

````python
simulation_time_ns = 100
simulation_time_ps = simulation_time_ns * 1000      # ns -> ps

rmsd_nm = 0.24
rmsd_angstrom = round(rmsd_nm * 10, 2)              # nm -> angstrom, rounded

print("Time:", simulation_time_ps, "ps")
print("RMSD:", rmsd_angstrom, "angstrom")
print("Is the RMSD below 0.3 nm?", rmsd_nm < 0.3)
````

Expected output: `Time: 100000 ps`, `RMSD: 2.4 angstrom`, `Is the RMSD below 0.3 nm? True`.

## Part 5 · Lists and indexing

📘 A **list** keeps several values in order, inside square brackets. It is how you will store a column of RMSD values.

- **Index** = position. Python starts counting at **0**, so the first item is `[0]`. (Think of the index as *"how many steps from the start"*.)
- **Negative index** counts from the end: `[-1]` is the last item.
- `len(x)` is the number of items; `max(x)` and `min(x)` are the largest and smallest.
- `x.append(v)` adds `v` at the end; `x[i] = v` replaces an item.
- **Slice** `x[1:4]` takes positions 1, 2 and 3 (the end position is *not* included).

▶️ **Worked example 5** (illustrative values)

````python
rmsd_values = [0.12, 0.15, 0.18, 0.16, 0.21]

print(rmsd_values[0])           # first item
print(rmsd_values[-1])          # last item
print(len(rmsd_values))         # how many items

rmsd_values.append(0.19)        # add a new observation at the end
rmsd_values[1] = 0.14           # change the second item
print(rmsd_values)
print(rmsd_values[1:4])         # a slice: items at positions 1, 2, 3
print(max(rmsd_values), min(rmsd_values))
````

Expected output: `0.12`, `0.21`, `5`, then `[0.12, 0.14, 0.18, 0.16, 0.21, 0.19]`, `[0.14, 0.18, 0.16]` and `0.21 0.12`.

⚠️ **Common mistake:** a list with 5 items has positions **0 to 4**. Asking for position 5 gives `IndexError: list index out of range`.
